```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_clean = pd.read_csv("../data/processed/Housing_clean.csv")
numeric_features = ["area", "bedrooms", "bathrooms", "stories", "parking"]
```

```python
# Histograms
df_clean[numeric_features].hist(bins=15, figsize=(15, 6))
plt.suptitle("Numeric Feature Distributions (Cleaned & Scaled)")
plt.savefig("../visualizations/final eda/num_feature_dist.png", dpi=300)
plt.show()
```

```python
# Boxplots to verify outlier removal
plt.figure(figsize=(15, 6))
sns.boxplot(data=df_clean[numeric_features])
plt.title("Boxplot of Scaled Numeric Features")
plt.savefig("../visualizations/final eda/num_boxplot_scaled.png", dpi=300)
plt.show()
```
